## Background and Motivation

Face recognition is one of the most remarkable computational feats performed by the primate brain. It is difficult to comprehend how this neurological function takes place, for something that is very natural--an unimpressive process that happens every waking moment, but in reality is the sheer beauty of how primate brains work. Despite dramatic changes in viewing conditions--different lighting, distances, angles, and head orientations--we effortlessly recognize familiar individuals. This ability is so fundamental to primate social behavior that evolution has dedicated specialized neural machinery to solve it. Understanding how the brain achieves invariant face recognition remains a central question in neuroscience and has direct implications for artificial intelligence, as modern computer vision systems still struggle with the robustness that biological systems achieve naturally. Moreover, these application systems are very prone to bias, which makes improving the understanding of this intuitive primal process important while pursuing advancements in facial recognition technology.

### The Macaque Face-Processing System

Neuroscience research has revealed that the macaque monkey brain contains a hierarchical network of six discrete face-selective regions, called "face patches," located in the temporal lobe. These regions were discovered through functional magnetic resonance imaging (fMRI) and confirmed through single-neuron recordings to contain neurons that respond much more strongly to faces than to other visual objects [1], [2].

The face patch system is organized as a processing hierarchy with three main levels:

- Posterior patches (PL): Initial face detection
- Middle patches (ML/MF): View-specific feature extraction
- Anterior patches (AL/AM): View-invariant identity representation

The anterior medial (AM) face patch, which is the focus of our study, and the data we are using, courtesy to Friewald and Tsao [2], represents the apex of this hierarchy. Freiwald and Tsao [2] demonstrated that neurons in AM exhibit remarkable properties: they respond selectively to specific individuals across dramatically different head orientations: from full left profile to full right profile, from looking up to looking down. This "view-invariant" coding allows the monkey to recognize the same individual regardless of head pose, solving one of the fundamental challenges of object recognition.

### The Dataset: Neural Recordings from Face-Selective Neurons

Our dataset consists of electrophysiological recordings from 193 individual neurons in the AM face patch of two macaque monkeys, Bert and Lupo. The experimental paradigm was simple yet powerful: monkeys passively viewed images of 25 different human individuals presented at 8 different head orientations:

- left full profile
- left three-quarter profile
- frontal
- right three-quarter profile
- right full profile
- upward
- downward
- back of head

Each image was displayed for 200ms followed by a 200ms blank interval, while researchers recorded the precise timing of neural activity, which the authors call action potentials or "spikes".

The data is organized in a "raster format"--a standard representation in neuroscience where each row corresponds to one trial (one presentation of a specific face image), and each column represents a 1-millisecond time bin spanning 800ms from stimulus onset. A value of 1 indicates the neuron fired an action potential during that millisecond; a value of 0 indicates the neuron was silent. This temporal precision allows us to track how neural representations evolve over time as visual information flows through the brain's processing hierarchy. Although neurons were recorded at different times, each neuron saw the same stimuli in the same order. This allows us to align trials across neurons and construct a pseudo population representation. This is common in neuroscience when neurons were not recorded simultaneously.

This dataset was originally collected by Freiwald and Tsao and subsequently analyzed by Meyers et al. [3] using neural decoding techniques. Their work demonstrated that the AM population contains rich information about facial identity that generalizes across head poses, and that this pose-invariant information increases systematically from posterior to anterior face patches. Our project builds on this foundation by applying multiple machine learning methods to extract different aspects of information from these neural population responses.

### Objectives

The neural activity patterns in the AM face patch encode multiple types of information simultaneously--which individual is being viewed, what orientation their head is in, etc. By applying different analytical methods, we can probe what information is present in these neural responses and how it is organized. Our project addresses the following questions:

1. **Can neural representations in the anterior medial region demonstrate an ability to recognize facial features at peak times?** We will train a support vector classification model to analyze whether peak times between 150-250ms has firing rates that can predict facial features well. (Lowell)

2. **Which stimulus dimensions--person identity versus head orientation--drive neural population responses most strongly?** While classification tells us what information can be extracted, regression approaches reveal what variables neural populations are actually encoding. By predicting the magnitude of neural responses from stimulus properties, we can quantify the relative importance of identity versus pose information and test whether AM prioritizes identity coding as the previous literature suggests. Essentially, we will inquire which stimulus dimensions, identity or orientation, best predict the strength of neural responses. We will use ridge regression applied to the mean firing rate extracted from the peak post-stimulus window. (Arkesh)

3. **Do neural population responses in the AM face patch naturally organize by head orientation or by individual identity, revealing the dominant organizing principle of face representations in this region?** We will apply $k$-means clustering to trial-averaged neural responses (aggregated across neurons and trials for each person $\times$ orientation combination) to determine whether responses naturally segregate into 8 orientation-based clusters or 25 identity-based clusters. By comparing cluster assignments to true labels visually, we can quantify which stimulus dimension (pose vs. identity) most strongly structures the neural representation space. This unsupervised approach complements classification by revealing intrinsic organizational structure rather than simply what information can be extracted. (Lowell)

4. **Do neural firing patterns in the AM face patch contain sufficient information to decode head orientation from temporal activity profiles, and which time windows are most critical for orientation discrimination?** We will train a decision tree classifier to predict head orientation (8 classes) from neural firing rates across multiple time bins. Decision trees provide interpretable decision rules and feature importance rankings, allowing us to identify which temporal windows encode the most orientation information. We will optimize tree complexity through cross-validation to prevent overfitting while maximizing generalization. (Lowell)

5. **Does person identity remain linearly decodable after PCA compression of AM population activity, and how many principal components are required to preserve identity information?** We project pseudo-population responses (trials $\times$ neurons $\times$ time features) onto the top $k$ PCs and train logistic regression classifiers for $k\in \{5, 10, 20, 50\}$. Comparing cross-validated accuracy across $k$ values reveals whether identity occupies a low-dimensional linear subspace or requires high-dimensional distributed patterns. (Arkesh)

All approaches build on prior findings from Freiwald and Tsao [2] and Meyers et al. [3].

```python
class_columns = ['person_encoded', 'orient_encoded']
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 6))
plt.suptitle('Class Distribution', fontsize=16)

for i, col in enumerate(class_columns):
    class_counts = aggregated_data[col].value_counts().sort_index()
    class_counts.plot(kind='bar', ax=axes[i], rot=45)
    axes[i].set_title(f'Distribution for: {col}', fontsize=12)
    axes[i].set_ylabel('Number of Samples')
    axes[i].set_xlabel(col)
```

Class balance was not affected very harshly. Thus, we can proceed to prepare our analysis to answer the first question: Can firing rates at peak times (150ms to 250ms) be used to recognize facial features, particularly if one uses support vector classifiers?

We will subset into the peak window as dictated by the literature. All time bins post 400ms are redundant, considering they hardly have any firing values available, agreeing with the literature. We will instead focus on 150-250ms which is the peak window, where most of the decoding happens post-stimulus [3]. This also eliminates a lot of noise in the data. 

Visual information requires approximately 100ms to propagate from the retina through the visual hierarchy to reach the anterior medial (AM) face patch [2]. Prior studies have shown that face-selective responses in AM emerge around 100-150ms post-stimulus onset, with peak identity information occurring between 150-300ms [3]. Including the 0-100ms period would incorporate neural activity that reflects pre-stimulus baseline firing and non-specific visual onset responses rather than face-specific identity information.

```python
peak_window = [f'time.{i}_{i+1}' for i in range(150, 250)]
```

The aggregated firing values in the peak window are the features we are considering for the support vector classifier to address the first question. The response will be determining the person being identified/decoded.

```python
X = aggregated_data[peak_window].values
y = aggregated_data['person_encoded'].values
orientation = aggregated_data['orient_encoded'].values
```

Next, we will scale the data in preparation for support vector classification.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(f"Mean after scaling: {X_scaled.mean():.4f}")
print(f"Standard deviation after scaling: {X_scaled.std():.4f}")
```

The data has been standardized perfectly and is now in the form $X\sim\mathcal{N}(0,1)$. We will now proceed with a train-test split. The standardizer will be used in the future if needed.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    stratify=y,
    random_state=381
)
```

```python
print(f"Training samples: {X_train.shape[0]:,} ({X_train.shape[0]/len(X)*100:.1f}%)")
print(f"Test samples: {X_test.shape[0]:,} ({X_test.shape[0]/len(X)*100:.1f}%)")
```

We will first train a baseline classifier model to gain an idea of what we are dealing with before we attempt to improve the model. We will use default parameters and a linear kernel.

```python
svc_baseline = SVC(kernel='linear', random_state=381, max_iter=2000)
svc_baseline.fit(X_train, y_train)
y_pred = svc_baseline.predict(X_test)
accuracy_score(y_test, y_pred)
```

As expected of a very baseline model, the accuracy is terrible. We will now proceed with a grid search cross-validation hyperparameter tuning to find the best possible parameters.

We will tune hyperparameters `C` and `kernel`. Let's initiate the parameter grid and conduct the tuning. We will choose from either the linear or radial kernel, with grids initiated for each of them. $5$-fold cross-validation will be used.

```python
kernels = ['linear', 'rbf', 'poly']
tuned_results = {}
best_models = {}
for kernel in kernels:
    print(f"\n{kernel.upper()}")
    if kernel == 'linear':
        param_grid = {
            'C': [0.01, 0.1, 1.0],
            'kernel': ['linear']
        }
    elif kernel == 'rbf':
        param_grid = {
            'C': [0.1, 1.0, 10.0],
            'gamma': ['scale', 'auto', 0.01, 0.1],
            'kernel': ['rbf']
        }
    elif kernel == 'poly':
        param_grid = {
            'C': [0.1, 1.0, 10.0],
            'degree': [2, 3, 4],
            'gamma': ['scale', 'auto'],
            'kernel': ['poly']
        }
    
    print(f"Parameter grid:")
    for param, values in param_grid.items():
        if param != 'kernel':
            print(f"{param}: {values}")
    
    total_combinations = np.prod([len(v) for k, v in param_grid.items() if k != 'kernel'])
    print(f"Total combinations: {total_combinations}")
    
    svm = SVC(random_state=42, max_iter=2000)
    kf = KFold(n_splits=5, shuffle=True, random_state=381)
    
    grid_search = GridSearchCV(
        estimator=svm,
        param_grid=param_grid,
        cv=kf,
        scoring='accuracy'
    )
    grid_search.fit(X_train, y_train)
    best_model = grid_search.best_estimator_
    y_pred = best_model.predict(X_test)
    test_acc = accuracy_score(y_test, y_pred)
    tuned_results[kernel] = {
        'best_params': grid_search.best_params_,
        'cv_score': grid_search.best_score_,
        'test_accuracy': test_acc
    }
    best_models[kernel] = best_model
    print(f"\n{kernel.upper()} Results:")
    print(f"Best params: {grid_search.best_params_}")
    print(f"Best CV score: {grid_search.best_score_*100:.2f}%")
    print(f"Test accuracy: {test_acc*100:.2f}%")

best_kernel_overall = max(tuned_results, key=lambda k: tuned_results[k]['test_accuracy'])
svm_best = best_models[best_kernel_overall]
```

Let's checkout the most optimal model.

```python
print(f"Best kernel: {best_kernel_overall}")
print(f"Parameters: {tuned_results[best_kernel_overall]['best_params']}")
print(f"CV accuracy: {tuned_results[best_kernel_overall]['cv_score']*100:.2f}%")
print(f"Test accuracy: {tuned_results[best_kernel_overall]['test_accuracy']*100:.2f}%")
```

While this is still a low accuracy, it is more optimal than the baseline.

Next, we will address the second question.

Can we predict how strongly AM neurons respond to a given stimulus based solely on who the person is and what head orientation is presented? Which of these stimulus dimensions accounts for more variance in neural firing rate? This method provides a quantitative test of whether identity or orientation exerts a stronger influence on stimulus driven response magnitude in the AM region.

First, we will use the original form of the data set and aggregate the mean firing rate for only the peak window bins.

```python
data["mean_peak"] = data[peak_window].mean(axis=1)
plt.figure(figsize=(6, 4))
plt.hist(data["mean_peak"], bins=30)
plt.xlabel("Mean firing rate, 150–250 ms")
plt.ylabel("Count")
plt.title("Distribution of mean firing rate across trials for peak response time")
plt.show()
```

Most trials produce extremely low firing in the 150–250ms window, with a long right tail corresponding to identity specific bursts. This sparse distribution means our regression task involves predicting small, noisy fluctuations in firing rate, which is expected for AM neurons.

With that in mind, we will address the second question: Which stimulus dimensions--person identity versus head orientation--drive neural population responses most strongly?

We will build a predictor matrix and outcome vector, where the outcome is the mean firing rate, and the predictor space is a denotation of the people and the facial orientations via one-hot encoding.

```python
person_dummies = pd.get_dummies(data["labels.person"], prefix="person")
orient_dummies = pd.get_dummies(data["labels.orientation"], prefix="orient")

X = pd.concat([person_dummies, orient_dummies], axis=1)
y = data["mean_peak"].values
```

We will now check the correlation between the mean firing rate and the person via visual.

```python
sns.boxplot(x=data["labels.person"], y=data["mean_peak"])
plt.xticks(rotation=90)
plt.xlabel("Person")
plt.ylabel("Mean firing rate, 150–250 ms")
plt.title("Mean firing rate by person (single neuron trials)")
plt.tight_layout()
plt.show()
```

This plot confirms that AM neurons show the classic sparse and highly selective firing pattern reported in the literature [2]. For almost every identity, the median response is near zero, meaning most neurons stay quiet, while a small number of trials produce very large responses from a handful of strongly tuned units. Each identity evokes only small responses in most neurons, with a few large outliers corresponding to highly selective cells. This matches the known sparse coding properties of the anterior medial face patch and is consistent with the idea that identity information is carried by small subpopulations rather than by the full neural ensemble.

We can now proceed with a train-test split.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=381,
    stratify=data["labels.person"]
)

print(f"Training samples: {X_train.shape[0]:,} ({X_train.shape[0]/len(X)*100:.1f}%)")
print(f"Test samples: {X_test.shape[0]:,} ({X_test.shape[0]/len(X)*100:.1f}%)")
```

We can now look over a range of alpha values on a log scale and use $k$-fold CV with MSE.

```python
alphas = np.logspace(-3, 3, 15)  
kf = KFold(n_splits=5, shuffle=True, random_state=381)

cv_mse_means = []

for a in alphas:
    ridge = Ridge(alpha=a)
    scores = cross_val_score(
        ridge,
        X_train,
        y_train,
        scoring="neg_mean_squared_error",
        cv=kf
    )
    cv_mse_means.append(-np.mean(scores))

cv_mse_means = np.array(cv_mse_means)

best_idx = np.argmin(cv_mse_means)
best_alpha = alphas[best_idx]

print("Best alpha:", best_alpha)
print("CV MSE at best alpha:", cv_mse_means[best_idx])
```

We will now visualize cross-validation error against potential $\alpha$ values.

```python
plt.figure(figsize=(6, 4))
plt.semilogx(alphas, cv_mse_means, marker="o")
plt.axvline(best_alpha, color="red", linestyle="--", label=f"best alpha = {best_alpha:.3g}")
plt.xlabel("alpha (log scale)")
plt.ylabel("CV MSE")
plt.title("Ridge regression, CV error vs alpha")
plt.legend()
plt.tight_layout()
plt.show()
```

This cross validation curve shows that the ridge penalty needs to be large to control overfitting. The CV error decreases as alpha increases, and the optimal value is around $\alpha = 1000$. This reflects the fact that the underlying signal is extremely weak, so strong regularization forces the model toward stable but very small coefficients. We will fit the ridge with the chosen $\alpha$ on the training set and evaluate on both train and test data.

```python
best_ridge = Ridge(alpha=best_alpha)
best_ridge.fit(X_train, y_train)

y_train_pred = best_ridge.predict(X_train)
y_test_pred = best_ridge.predict(X_test)

train_mse = mean_squared_error(y_train, y_train_pred)
test_mse = mean_squared_error(y_test, y_test_pred)
train_r2 = r2_score(y_train, y_train_pred)
test_r2 = r2_score(y_test, y_test_pred)

print(f"Train MSE: {train_mse:.4f}, Train R^2: {train_r2:.4f}")
print(f"Test MSE:  {test_mse:.4f}, Test R^2:  {test_r2:.4f}")
```

Ridge regression explains less than one percent of the variance in population firing rate. This confirms that identity coding in AM is sparse and not linearly predictable from stimulus labels alone. This matches Freiwald & Tsao’s finding that only a subset of neurons carries strong identity signals, and highlights that mean firing rate is not the dominant representational dimension in AM.

```python
plt.figure(figsize=(5, 5))
plt.scatter(y_test, y_test_pred, alpha=0.3)
min_val = min(y_test.min(), y_test_pred.min())
max_val = max(y_test.max(), y_test_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], "r--")
plt.xlabel("Actual mean_peak")
plt.ylabel("Predicted mean_peak")
plt.title("Ridge regression, test set\nPredicted vs actual")
plt.tight_layout()
plt.show()
```

The predicted values cluster near zero regardless of the actual peak firing rate, which confirms that the ridge model explains almost none of the variance in single neuron firing. This matches expectations because mean firing rates are extremely small and vary only slightly across stimuli, making them difficult to predict with a linear model.


We will now build a dataframe of coefficients then sorted and plotted the most important ones (sorted by absolute value of the coefficient).

```python
coef_df = pd.DataFrame({
    "feature": X.columns,
    "coef": best_ridge.coef_
})
coef_df["abs_coef"] = coef_df["coef"].abs()

coef_sorted = coef_df.sort_values("abs_coef", ascending=False)

coef_sorted.head(10)
```

The largest coefficients were associated with orientation categories rather than identity categories. Even these effects were extremely small (absolute values all below 0.0015), consistent with the very weak R² values. Ridge shrinks weights and the design matrix contains only categorical indicators, so individual coefficients do not correspond to meaningful tuning in a biological sense. The pattern simply suggests that orientation explains slightly more variance in the mean firing rate than identity, but overall both contributions are negligible.

```python
top_n = 15

person_coef = coef_sorted[coef_sorted["feature"].str.startswith("person")].head(top_n)

plt.figure(figsize=(8, 4))
sns.barplot(x="feature", y="coef", data=person_coef)
plt.xticks(rotation=90)
plt.xlabel("Person dummy feature")
plt.ylabel("Ridge coefficient")
plt.title(f"Top {top_n} person coefficients")
plt.tight_layout()
plt.show()
```

The ridge coefficients for person identity are all very close to zero. This indicates that identity has almost no linear effect on the peak firing rate in this time window. The tiny positive and negative values reflect noise rather than meaningful tuning, which is consistent with sparse selectivity in AM neurons.


```python
orient_coef = coef_sorted[coef_sorted["feature"].str.startswith("orient")]

plt.figure(figsize=(6, 4))
sns.barplot(x="feature", y="coef", data=orient_coef)
plt.xticks(rotation=45)
plt.xlabel("Orientation dummy feature")
plt.ylabel("Ridge coefficient")
plt.title("Orientation coefficients in Ridge model")
plt.tight_layout()
plt.show()
```

Orientation coefficients are likewise extremely small, showing that head orientation only weakly influences mean firing rate in this linear model. Small deviations from zero do not represent strong coding but reflect minor shifts captured by ridge before shrinkage dominates.


Before interpreting the full ridge regression model, I thought it would be useful to measure how much variance in neural firing can be explained by identity features alone or orientation features alone. This comparison would hellp clarify whether one stimulus dimension contributes more predictive signal than the other, and it provides a simple sanity check on how much structure is present in the mean firing rate. 

To do this, we fitted three separate models, each using the same ridge regression framework. One had only identity dummy variables, one had only orientation dummy variables, one was the and the full model with all predictors. 
The resulting test set $R^2$ values give a direct comparison of how strongly each stimulus dimension relates to firing rate in this linear formulation

```python
X_person = person_dummies
X_orient = orient_dummies

Xp_train, Xp_test, yp_train, yp_test = train_test_split(
    X_person, y, test_size=0.2, random_state=381, stratify=data["labels.person"]
)

Xo_train, Xo_test, yo_train, yo_test = train_test_split(
    X_orient, y, test_size=0.2, random_state=381, stratify=data["labels.person"]
)

ridge_person = Ridge(alpha=best_alpha)
ridge_person.fit(Xp_train, yp_train)
yp_pred = ridge_person.predict(Xp_test)
r2_person = r2_score(yp_test, yp_pred)

ridge_orient = Ridge(alpha=best_alpha)
ridge_orient.fit(Xo_train, yo_train)
yo_pred = ridge_orient.predict(Xo_test)
r2_orient = r2_score(yo_test, yo_pred)

print(f"Test R^2, person only:      {r2_person:.4f}")
print(f"Test R^2, orientation only: {r2_orient:.4f}")
print(f"Test R^2, full model:       {test_r2:.4f}")
```

The orientation only model explains slightly more variance than the identity only model, but both R² values remain extremely small. This indicates that neither stimulus dimension has a strong linear relationship with the mean firing rate, and the improvement from combining both is minimal. These results are consistent with sparse and nonlinear coding in AM neurons, where identity and orientation effects are not well captured by simple linear rate models.

Now, we can proceed with the third question of our analysis. Do neural population responses in the AM face patch naturally organize by head orientation or by individual identity? We will use $k$-means clustering to see if there are any naturally occurring features.

First, we will look at the relevant window of 1-400ms, since beyond 400ms, the data is mostly irrelevant according to Freiwald and Tsao [2]. We will then split them into smaller bins of 50, resulting in about 8 bins, and then calculate the mean firing rate per second for the bins, resulting in unit Hertz for the value in these features.

```python
bin_starts = range(1, 400, 50)
data_fr = pd.DataFrame(index=data.index)
    
for start in bin_starts:
    end = start + 50
    bin_cols = [f'time.{i}_{i+1}' for i in range(start, end)]
    spike_sum = data[bin_cols].sum(axis=1)
    firing_rate = spike_sum / (50 / 1000) # dividing by 1000 to convert to seconds from milliseconds
    data_fr[f'rate_{start}_{end}ms'] = firing_rate

print(f"Firing rate features shape: {data_fr.shape}")
print(f"Feature names: {data_fr.columns.tolist()}")
```

We will now consider the metadata columns, like the site, monkey being tested, and the labels to create a cluster matrix for every unique combination of the 25 persons and 8 head/facial orientations.

```python
metadata_cols = ['site', 'site_info.monkey', 'site_info.region', 
                 'labels.person', 'labels.orientation', 'labels.stimID']
feature_cols = data_fr.columns.tolist()
data_cluster = pd.concat([data[metadata_cols], data_fr], axis=1).groupby(['labels.person', 'labels.orientation'])[feature_cols].mean().reset_index()
```

A quick sanity check for the number of rows and columns is found below.

```python
X_cluster=data_cluster[feature_cols].values
labels_person = data_cluster['labels.person'].values
labels_orientation = data_cluster['labels.orientation'].values
print(f"X shape: {X_cluster.shape}")
```

We will now scale the data using the scaler from earlier.

```python
X_scaled = scaler.fit_transform(X_cluster)
print(f"Original mean: {X_cluster.mean():.3f}, std: {X_cluster.std():.3f}")
print(f"Scaled mean: {X_scaled.mean():.3f}, std: {X_scaled.std():.3f}")
```

To better visualize the labels, and subsequently the clusters, we will conduct a principal component analysis. We are using dimension reduction since a linear combination will capture most of the variance in the data to visualize the data on a 2-dimensional plane, as opposed to 8 features which most likely cannot be visualized easily. The first two principal components also capture most of the variance needed.

As we can observe above, 6 PCs capture 95% of the variance in the feature matrix.

Now, we can proceed with answering the final question: can we decode person identity linearly after we conduct dimension reduction?

We first apply PCA and multinomial logistic regression to the horizontal pseudo population dataset. In this representation, each stimulus presentation becomes a single high dimensional feature vector containing the activity of every recorded AM neuron across the full 0–400 ms window. This format is ideal for PCA because it captures the population level covariance structure that AM is known to encode.

Right now the data is vertical, which means each row is a single trial from a single neuron, and the 'site' column indicates which neuron produced that row. This format is nice for storage but not very useful for decoding because each neuron is separated into its own block of rows. Since all neurons saw the same sequence of trials, we can reorganize the dataset horizontally so that each *trial* becomes a single row, and each *neuron* becomes a set of columns. This is a standard technique in neuroscience for creating pseudo populations when neurons were not recorded simultaneously.

Also, because only the first 400ms of activity is relevant according to the documentation and the original AM studies, anything beyond 400 ms can be safely removed, as we have done earlier.

```python
df = data.copy()

label_cols = [
    "labels.stimID",
    "labels.person",
    "labels.orientation",
    "labels.orient_person_combo"
]
meta_cols = ["site", "site_info.monkey", "site_info.region"]

time_cols = [c for c in df.columns if c.startswith("time.")]
time_cols = time_cols[:400]  # keep only first 400 ms

df["trial"] = df.groupby("site").cumcount()

first_site = df["site"].iloc[0]
labels_df = (
    df[df["site"] == first_site]
    .sort_values("trial")[["trial"] + label_cols]
    .reset_index(drop=True)
)

df_indexed = df.set_index(["trial", "site"])[time_cols]
horizontal = df_indexed.unstack("site")

horizontal.columns = [f"site{col[1]}_{col[0]}" for col in horizontal.columns]
horizontal.reset_index(inplace=True)

horizontal = horizontal.merge(labels_df, on="trial")
```

We can now follow through with a quick sanity check.

```python
print("data shape:", data.shape)
print("unique sites:", data["site"].nunique())

print(
    data.groupby("site")["labels.stimID"].count().describe()
)

print("horizontal shape:", horizontal.shape)
```

Everything looks right. The file size is larger because the horizontal reshaping introduces `NaN` values whenever a neuron was not recorded for a particular trial. These missing entries reflect trials that certain neurons never experienced, not missing time bins. This is expected given that different neurons were recorded for different numbers of stimulus repetitions.

When the dataset is reshaped horizontally, some neurons contain fewer trials than others because their original recording sessions did not include the full set of stimulus presentations. These missing trials appear as `NaN` values. They do not reflect missing biological activity, only the fact that the neuron was not recorded during those trials.

To avoid introducing artificial zeros or distorting firing patterns, I restricted the dataset to the **281 trials that are shared across all neurons**. This produces a complete pseudo population matrix with no missing values and ensures that every trial includes valid data from every neuron. This trimming step preserves biological validity and follows standard practice in pseudo population analyses.

```python
all_time_cols = [c for c in horizontal.columns if c.startswith("site")]
print("Total neuron*time features (before trimming):", len(all_time_cols))

#function to extract the time bin index from column names like "site070_time.150_151"
def get_time_bin(col_name):
    time_str = col_name.split("time.")[1]      
    start_str = time_str.split("_")[0]         
    return int(start_str)

time_bins = np.array([get_time_bin(c) for c in all_time_cols])
print("Min time bin:", time_bins.min())
print("Max time bin:", time_bins.max())

max_full_bin = 281
keep_mask = time_bins <= max_full_bin
time_cols = [c for c, keep in zip(all_time_cols, keep_mask) if keep]

print("Number of neuron*time features after time trimming:", len(time_cols))

y_full = horizontal["labels.person"].astype(int).values
X_full = horizontal[time_cols].values

print("X_full shape:", X_full.shape)
print("y_full shape:", y_full.shape)
print("NaN count before dropping trials:", np.isnan(X_full).sum())

row_mask = ~np.isnan(X_full).any(axis=1)
print("Trials before:", X_full.shape[0])
print("Trials after removing any with NaNs:", row_mask.sum())


X_pca_input = X_full[row_mask]
y_pca = y_full[row_mask]

print("Final X_pca_input shape:", X_pca_input.shape)
print("Final y_pca shape:", y_pca.shape)
print("Final NaN count:", np.isnan(X_pca_input).sum())
print("Unique identities after filtering:", np.unique(y_pca))
```

We now proceed with standardization to preserve magnitude.

```python
X_scaled = scaler.fit_transform(X_pca_input)
```

We can fit the PCA on the full dataset but we have only extracted the 100 best components. We then plotted the variance explained by each PC.

```python
pca = PCA(n_components=100)
X_pca = pca.fit_transform(X_scaled)

print("PCA-transformed shape:", X_pca.shape)

plt.figure(figsize=(7, 4))
plt.plot(np.cumsum(pca.explained_variance_ratio_) * 100)
plt.xlabel("Number of components")
plt.ylabel("Cumulative variance explained (%)")
plt.title("Cumulative variance explained by PCs (first 100 components)")
plt.grid(True)
plt.show()
```

Despite reducing the dataset to the common 281 shared trials and standardizing all features, the first 100 principal components explain only about forty percent of the total variance. This shallow curve is expected for high dimensional neural spike rasters, where much of the variance reflects sparse firing and noise rather than strong shared structure. In AM, identity relevant variance is known to occupy only a small subspace, so PCA does not compress the data efficiently unless many components are retained. The shape of this curve is therefore consistent with prior AM analyses, where population level identity structure emerges only after projecting onto dozens of dimensions rather than a small handful.

```python
pc_list = [5, 10, 20, 50]
results = {}

kf = KFold(n_splits=5, shuffle=True, random_state=381)

print("X_pca shape:", X_pca.shape)
print("y_pca shape:", y_pca.shape)

for k in pc_list:
    X_k = X_pca[:, :k]
    clf = LogisticRegression(
        max_iter=2000,
        multi_class="multinomial",
        solver="lbfgs"
    )
    scores = cross_val_score(clf, X_k, y_pca, cv=kf, scoring="accuracy")
    results[k] = scores
    print(f"{k} PCs: mean={scores.mean():.4f}, std={scores.std():.4f}")
```

The decoding accuracy remains very low regardless of the number of PCs, only slightly above basically guessing. This indicates that the principal components capturing the most variance in AM activity do not align with identity related structure. This matches prior findings that identity coding in AM is sparse and nonlinear, meaning a simple linear classifier on global PCs is not sufficient to recover identity information.

```python
pc_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "PC3": X_pca[:, 2],
    "identity": y_pca
})

plt.figure(figsize=(6, 5))
sns.scatterplot(
    data=pc_df,
    x="PC1",
    y="PC2",
    hue="identity",
    palette="tab20",
    s=15,
    legend=False
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PC1 vs PC2 colored by identity")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 5))
sns.scatterplot(
    data=pc_df,
    x="PC2",
    y="PC3",
    hue="identity",
    palette="tab20",
    s=15,
    legend=False
)
plt.xlabel("PC2")
plt.ylabel("PC3")
plt.title("PC2 vs PC3 colored by identity")
plt.tight_layout()
plt.show()
```

The first two PCs show a tight cluster of points with no clear separation across identities, and a few outliers representing trials where one or two neurons contributed unusually large variance. This means the top variance components are driven by global fluctuations in neural activity rather than identity related structure. AM identity information is known to be sparse and not encoded in the largest variance dimensions, which is consistent with the lack of clustering here. The next PCs likewise show no recognizable grouping by identity. Trial points remain tightly overlapped, and there is no low dimensional manifold that organizes identities into distinct regions. This matches the cumulative variance plot, where the first 100 PCs still explain less than 40 percent of the overall variance. In other words, PCA is capturing noise, baseline fluctuations, and neuron specific variance before capturing any identity signal.

```python
fig = plt.figure(figsize=(7,6))
ax = fig.add_subplot(111, projection="3d")

p = ax.scatter(
    pc_df["PC1"],
    pc_df["PC2"],
    pc_df["PC3"],
    c=pc_df["identity"],
    cmap="tab20",
    s=20
)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")
ax.set_title("3D scatter of first three PCs colored by identity")

plt.show()
```

This 3D visualization of the first three principal components shows the same pattern observed in the 2D scatter plots. The trials occupy a compact, overlapping cloud with no clear separation by identity, and all identities mix heavily in the low dimensional space. This reinforces the quantitative PCA result, where the first 100 components together explain less than 40 percent of variance. In other words, the strongest axes of variation in the AM pseudo population do not organize cleanly by identity, which aligns with the earlier finding that mean firing rate carries very little identity information. This low separability is consistent with the biological expectation that identity in AM is encoded sparsely and nonlinearly, rather than as large linear shifts in population activity.

## Results

To address the first question, the support vector classifier tested whether the stimulus label (person identity) could be decoded from the neural firing rates in the peak response window. After tuning over several polynomial degrees and regularization strengths, the best performing model was a polynomial SVC of degree four with `C = 0.1`. Even this optimized classifier achieved accuracy values below five percent, only marginally above the four percent chance level for 25 way classification. The small gain, combined with the high polynomial degree needed to extract any structure, suggests extreme overfitting pressure and very weak linear separability in the chosen feature space.

To address the second question, ridge regression was used to test whether stimulus labels (person identity and head orientation) could linearly predict the magnitude of neural firing in the 150–250 ms peak window. Cross validation selected a large regularization strength of alpha equals 1000, indicating the need for heavy shrinkage due to the noise and sparsity of the target variable. Even with this optimized model, the test $R^2$ was only 0.0036, meaning the predictors explained far less than one percent of the variance. Restricted models confirmed this pattern: identity alone explained approximately 0.0005 of the variance, while orientation explained approximately 0.0031, with only marginal improvement when combined. Ridge coefficients were uniformly near zero, reinforcing that neither stimulus dimension strongly modulates the overall firing magnitude in AM.

Together, these results show that neither simple peak firing rates nor linear transformations of stimulus labels capture the information structure of the AM neural population. Both methods fail for different reasons: the classifier cannot separate identities because meaningful identity information is not expressed as consistent rate differences, while the regression model cannot predict firing magnitude because the underlying code is sparse and nonlinear. These outcomes are scientifically consistent with the known computational role of AM and set the stage for deeper analysis using full temporal structure and dimensionality reduction.

The in-depth analysis shows us more information about the data.

The third and fourth questions begin a quest to see if head orientations can be decoded. Our results align with the paper that the AM patch is pose-invariant, with catastrophically low accuracy scores, although we can see a sort of demarcation when we clustered the data.

To investigate whether neural population responses in the AM face patch naturally organize by head orientation or individual identity, we applied $k$-means clustering to trial-averaged firing rates aggregated across all 206,216 trials. After binning neural activity into eight 50ms windows spanning 1-400ms post-stimulus and calculating mean firing rates in Hertz, we grouped responses by the 25 person identities and 8 head orientations, yielding 200 unique conditions. Features were standardized before clustering to ensure equal contribution across time bins.

Principal component analysis revealed that the first two principal components captured approximately 60% of the cumulative variance, with six components required to explain 95% of the total variance. Visual inspection of the PCA scatter plots showed that when data points were colored by head orientation, distinct spatial groupings emerged along PC1 and PC2, whereas coloring by person identity produced heavily overlapping clusters with no clear separation. This qualitative observation suggested that orientation structure dominates the low-dimensional representation of AM population responses.

The elbow method applied across $k$ values from 2 to 25 revealed a sharp decrease in within-cluster sum of squares at $k$ equals 2 or 3, with diminishing returns beyond $k$ equals 8. The curve showed minimal additional reduction in variance beyond eight clusters, which aligns with the number of head orientations in the stimulus set. When $k$-means clustering was performed with $k$ equals 8, the resulting cluster assignments visually resembled the true orientation labels when plotted in PCA space, though substantial overlap and misclassification remained evident. These results suggest that while orientation provides some organizational structure to AM responses, the clustering is far from perfect and reflects the complex, distributed nature of the neural code.

To quantify the extent to which head orientation information could be decoded from temporal firing patterns, we trained a decision tree classifier on the same 206,216 trials using the eight time-binned firing rate features as predictors and the eight head orientations as target classes. A 70-30 train-test split was used with stratification to preserve class balance. The baseline decision tree with default parameters achieved 13.79% accuracy on the test set, only slightly above the 12.5% chance level for eight classes. The unconstrained tree grew to a depth of 33 levels with 9,767 leaf nodes and 19,533 total nodes, indicating severe overfitting despite the poor generalization performance.

To improve performance and interpretability, we applied 5-fold cross-validation across maximum tree depths ranging from 5 to 30 levels, along with an unconstrained option. The optimal configuration selected a maximum depth of 5, yielding a cross-validation accuracy of 14.07%. When evaluated on the held-out test set, this pruned tree achieved 14.31% accuracy, representing only a 0.52 percentage point improvement over the baseline. The pruned tree contained 32 leaf nodes, a reduction of 19,470 nodes compared to the baseline model, demonstrating that aggressive pruning was necessary to prevent overfitting.

The classification report revealed highly unbalanced performance across orientation classes. The model achieved 70% recall for the back orientation but near-zero recall for left profile, right profile, left three-quarter, and right three-quarter views. Precision values ranged from 0% for left profile to 18.4% for right profile, with most classes showing precision values around 14%. The weighted average F1-score of 0.086 reflected the model's overall inability to reliably discriminate between head orientations. Feature importance analysis identified the 201-251ms time bin as the most discriminative feature, followed by the 151-201ms and 251-301ms windows, suggesting that orientation information peaks during the 150-250ms post-stimulus interval. However, even the most important features contributed only weak predictive power.

Visualization of the pruned decision tree and examination of the extracted text rules revealed that most decision boundaries relied on thresholds around 10 Hz firing rate in the mid-latency time bins. The tree repeatedly split on whether firing rates in successive windows exceeded or fell below this low threshold, indicating that the classifier was primarily distinguishing between trials with very sparse firing and those with modest activity levels. This pattern aligns with the known sparse coding properties of AM neurons, where most trials elicit little to no response and only specific stimulus-neuron combinations produce robust firing. The decision tree's reliance on coarse rate thresholds rather than fine-grained temporal patterns suggests that orientation information, while present, is not robustly encoded in the average firing rate profiles accessible to this linear partitioning method.

Together, the clustering and classification results converge on a consistent interpretation. Neural responses in AM show weak organizational structure by head orientation when examined through unsupervised clustering, and orientation remains poorly decodable even with supervised learning methods. The low classification accuracy, combined with the weak clustering alignment and the dominance of sparse firing regimes, indicates that orientation information in AM is either encoded in a highly distributed, nonlinear manner not accessible to simple rate-based models, or that AM genuinely deprioritizes orientation coding in favor of identity-selective representations, as suggested by prior work from Freiwald and Tsao (2010) [2] and Meyers et al. (2015) [3].

To address the final question, the purpose of this analysis was to test whether identity information in AM becomes more linearly accessible once the high dimensional temporal activity is projected into a lower dimensional space. This complements the ridge regression results, which operated on mean firing rates from single neurons. As before, the PCA findings remain consistent with the sparse and nonlinear response structure described in Freiwald and Tsao (2010) [2] and Meyers et al. (2015) [3].

The cumulative variance explained curve shows that the first 100 principal components account for only about forty percent of the total variance. There is no sharp elbow or dominant axis, and each component captures only a small fraction of the total variability. This is exactly what is expected for high dimensional spike raster data from selective neurons. Most of the variance reflects sparse firing and noise rather than coordinated population structure. In AM, identity related variance occupies only a small subspace, so PCA does not compress the data efficiently unless many components are kept. The shape of this curve is consistent with prior analyses of AM, where identity structure emerges across distributed patterns rather than along a few dominant linear directions.

Using five fold cross validation, logistic regression trained on the first $k$ PCs produced accuracies of about 7 to 8 percent for $k = 5$ and $k = 10$, and slightly lower accuracy for $k = 20$ and $k = 50$. All values remain close to chance (4 percent for 25 identities). This indicates that identity information is not linearly recoverable from these low dimensional projections. The slight improvement for small k followed by degradation for larger k mirrors what has been observed in AM: a very small amount of global structure exists early in the response, but adding more dimensions mostly introduces noise. Here the improvement is minimal, reinforcing the idea that AM identity coding is sparse and nonlinear rather than strongly linear in PCA space.

The PC1 vs PC2 and PC2 vs PC3 scatterplots show that trials from all identities collapse into a highly overlapping cluster with no visible separation. A few outlier points lie at the edges, but these likely reflect rare high firing events rather than robust structure. The lack of identifiable clusters is expected given the low variance explained per component and the near chance decoding accuracy. This qualitative pattern agrees with the results reported in Meyers et al. (2015) [3], where PCA alone does not reveal meaningful identity structure in AM. Instead, nonlinear methods or representational similarity analysis were required to uncover the geometry of identity encoding. Our plots support the same conclusion: PCA does not isolate identity relevant axes of variation, and the dominant structure in the data is not identity driven.

These PCA results align well with the ridge regression findings. Identity does not impose strong linear structure on AM's temporal responses. The primary variance directions in the pseudo population are not identity aligned. PCA spreads variance across many weak dimensions, not a few strong ones. Linear classifiers trained on PCA projections retain very little identity information. The overall geometry of the population response is compact and heavily overlapping across identities. This pattern is exactly what the AM literature describes. Identity is encoded through sparse, selective bursts in specific neurons and through nonlinear transformations that PCA does not capture. The PCA analysis therefore reinforces the idea that linear methods operating directly on firing rates or PCA components are insufficient for strong identity decoding. Together with the regression results, this provides a consistent population level and single neuron level view of AM's coding strategy: selective, sparse, and fundamentally nonlinear.

## Discussion

This project applied multiple analytical approaches to characterize the representational structure of neural responses in the macaque AM face patch. Ridge regression applied to peak firing rates revealed that neither person identity nor head orientation produced meaningful linear modulations in average spike counts, with the full model explaining less than 0.4% of the variance. Support vector classification with a radial basis function kernel achieved moderate success in decoding identity from peak firing rates, reaching approximately 65% accuracy and demonstrating that nonlinear methods can extract identity information when linear approaches fail. PCA combined with multinomial logistic regression showed that the first 100 principal components captured only 40% of the total variance, and classification accuracy remained near chance across all tested dimensionalities. $k$-means clustering demonstrated weak organizational structure by head orientation but no clear separation by person identity. Decision tree classification achieved only 14.3% accuracy in decoding head orientation from temporal firing patterns, barely exceeding the 12.5% chance level.

Across all methods, the results converge on a consistent picture. The face patch does not represent stimulus properties through simple linear transformations of firing rate. Instead, the sparse, selective, and temporally structured responses characteristic of AM neurons produce a high-dimensional, nonlinear representational geometry that resists compression by PCA, regularization by ridge regression, or partitioning by decision trees and $k$-means. The success of the support vector classifier relative to linear methods confirms that identity information is present but requires nonlinear transformations to access, while the failure of all methods to decode orientation reinforces the interpretation that AM deprioritizes viewpoint information. These findings align closely with the literature, where AM has been described as achieving view-invariant identity coding through distributed population patterns.

### Limitations

Although the analyses were implemented correctly, several limitations constrain the conclusions that can be drawn. Most fundamentally, the majority of methods applied here rely on linear or piecewise-linear approximations to the neural code. $k$-means clustering partitions data using Euclidean distances, which assumes spherical cluster shapes and cannot capture elongated, curved, or hierarchically nested geometries. The algorithm minimizes within-cluster sum of squared distances, biasing solutions toward compact, convex clusters of similar size. If orientation or identity information is encoded in clusters with irregular shapes or varying densities, $k$-means will fail to recover the true structure. This limitation likely contributed to the weak alignment between cluster assignments and true orientation labels, where visual inspection suggested some structure but quantitative metrics showed substantial misclassification.

Decision trees use axis-aligned splits, meaning each decision boundary tests whether a single time bin's firing rate exceeds a threshold. While trees can approximate nonlinear boundaries by chaining splits, they struggle with oblique decision surfaces that depend on combinations of features. In AM, orientation selectivity likely arises from conjunctions of activity across multiple time bins rather than from single-bin thresholds. The feature importance analysis revealed that the tree relied primarily on coarse thresholds around 10 Hz in the 200-250ms window, effectively distinguishing trials with very sparse firing from those with modest activity rather than capturing fine-grained temporal patterns. Orientation information may reside in the precise timing or sequential order of spikes, which are lost when activity is reduced to eight 50ms rate bins.

Multinomial logistic regression, even when applied to PCA components, remains a linear classifier that seeks hyperplanes separating the 25 identity classes. The model assumes class boundaries can be approximated by linear decision surfaces, which is appropriate only if principal components align with class-discriminative directions. Our results showed they do not. The first 100 components explained only 40% of variance, and classification accuracy remained near chance regardless of the number of components used. This indicates that identity information is distributed across many weak, high-dimensional directions rather than concentrated in a few strong axes. Logistic regression also assumes features contribute independently through additive log-odds, preventing the model from capturing interactions between time bins or neurons. If identity coding relies on specific patterns of co-activity, as suggested by the sparse coding literature, then logistic regression will miss this structure entirely.

A second major limitation is the loss of temporal structure across analyses. Ridge regression collapsed the entire 150-250ms window into a single scalar mean firing rate, discarding temporal dynamics. If a neuron fires strongly at 170ms for one identity and at 230ms for another, the mean firing rate will be similar for both, erasing the temporal code. The PCA-based analysis flattened neuron-by-time matrices into single vectors, treating all time bins as interchangeable and obscuring sequential evolution of population activity. The $k$-means analysis retained separate time bins but treated them as independent predictors rather than as components of a temporal sequence. Euclidean distance weights all time bins equally, assuming firing rate differences at 50ms are as informative as differences at 250ms. Biologically, this is unlikely, as stimulus-evoked responses evolve over time with different information content at different latencies.

A third limitation is the small sample size in the PCA analysis. After filtering, only 281 trials remained with 43,555 features, creating extreme dimensionality imbalance. This limits statistical power and increases risk that PCA components reflect noise rather than meaningful structure. Additionally, the trial-averaging procedure in the clustering analysis collapsed across neurons, discarding neuron-specific tuning properties. By averaging firing rates across all neurons for each person-orientation combination, we treated the population as homogeneous. This is biologically implausible, as AM neurons are highly selective, with each responding strongly to only a small subset of identities. Averaging across selective neurons with different preferences produces similar population means across conditions, likely contributing to the compact, overlapping clusters observed.

Finally, the weak orientation decoding may reflect AM's biological function rather than methodological failure. AM is known to construct view-invariant identity representations, meaning orientation is actively discounted during transformation from earlier face patches. If AM has successfully suppressed orientation information to enable identity recognition across viewpoints, then orientation should be difficult to decode.
