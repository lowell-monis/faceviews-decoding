## Background and Motivation

### **Dataset**

We will use the **Freiwald Tsao Face View AM dataset**, which records neural activity from macaques while they viewed faces of **25 individuals** across **8 head orientations**.  
Our goal is to **decode the visual stimulus** (face identity or head orientation) from these neural recordings.

**Tasks:**

- Decode **face identity** using:

  - Quadratic Discriminant Analysis (QDA)
  - K Nearest Neighbors (KNN)
- Use **cross validation** to estimate test error and compare models.
- Optionally apply **Ridge** or **Lasso** to identify neurons that matter most.

**Tasks:**

- Decode **head orientation** using:
  - Logistic Regression
  - LDA
  - QDA
  - KNN
- Run **unsupervised learning** to explore structure in the neural data:
  - **PCA** to reduce dimensionality and visualize major axes of variation
  - **Clustering** (K means or Hierarchical) to see if trials group by identity or orientation

**Deliverables:**

- Accuracy comparison for orientation decoding  
- PCA visualizations and variance explained
